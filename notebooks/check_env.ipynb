```python

import sys
from pathlib import Path

import pandas as pd
import sklearn
import IPython
import notebook
import ipykernel

project_root = Path.cwd()

# اگر نوت‌بوک از پوشه notebooks اجرا شود،
# ریشه پروژه یک سطح بالاتر است.
if project_root.name == "notebooks":
    project_root = project_root.parent

print("Python executable:", sys.executable)
print("Python version:", sys.version.split()[0])
print("Project root:", project_root)
print("pandas:", pd.__version__)
print("scikit-learn:", sklearn.__version__)
print("IPython:", IPython.__version__)
print("Notebook:", notebook.__version__)
print("ipykernel:", ipykernel.__version__)

required_dirs = [
    "data",
    "data/raw",
    "data/processed",
    "notebooks",
    "src",
    "tests",
    "reports",
]

for directory in required_dirs:
    exists = (project_root / directory).is_dir()
    print(f"{directory}: {'OK' if exists else 'MISSING'}")

assert ".venv" in sys.executable, (
    "The notebook is not using the project's virtual environment."
)

assert all(
    (project_root / directory).is_dir()
    for directory in required_dirs
), "One or more expected project directories are missing."

print("\nEnvironment check passed.")
```
